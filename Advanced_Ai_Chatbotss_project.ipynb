```
from decorator import append
from contextlib import contextmanager
from transformers import pipeline

agent = pipeline("text-generation", model="TinyLlama/TinyLlama-1.1B-chat-v1.0")
memory = []
memory_categories = {
    "personal": [],
    "questions": [],
    "facts": [],
    "tasks": []
}

def clean(text):
    text = text.strip()
    stop_tokens = [
        "User:",
        "AI Agent",
        "Human",
        "Conversation"
        # "I" was removed as it seemed like a typo or extraneous character
    ]

    for token in stop_tokens:
        text = text.split(token)[0]

    words = text.split() # Fixed typo: slip() to split()
    cleaned_words = []

    for i, word in enumerate(words):
        # Fixed: lower() should be applied to word, and correct comparison
        if i > 0 and word.lower() == words[i - 1].lower():
            continue

        cleaned_words.append(word) # Fixed: words to word
    return " ".join(cleaned_words).strip() # Join cleaned words and return

def categorize(user_input):
    text = user_input.lower()

    if any(word in text for word in ["name", "age", "live", "from"]):
        return "personal"

    elif any(word in text for word in ["what", "why", "how", "who", "when"]):
        return "questions"

    elif any(word in text for word in ["fact", "remember", "information", "data", "report"]):
        return "facts"

    else:
        return "tasks"

def memory_context():
    context = ""

    for category, items in memory_categories.items():
        if items:
            context += f"\n{category.upper()} MEMORY:\n" # Fixed typo: upger() to upper()
            # Changed to get last 3 items if available, otherwise fewer
            for item in items[-3:]:
                context += f"{item}\n" # Fixed typo: /n to \n

    return context # Fixed: returning context string instead of contextmanager

def ai_agent():
    print("\n===========================\n ")
    print("\n===   Smart AI Agent     ===\n ")
    print("Type '== Exit ==' to stop")

    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print('Agent:  Goodbye!')
            break

        category = categorize(user_input)
        memory_categories[category].append(user_input) # Fixed typo: categories to category
        memory.append(f"User: {user_input}")
        recent_history = "\n".join(memory[-6:])
        memory_context_str = memory_context() # Fixed: calling the function and assigning to a variable

        prompt = f"""
You are a helpful, AI Assistance

RULES:
- Provide short concise questions
- stay relevant

MEMORY:
{memory_context_str}

RECENT CONVERSATION:
{recent_history}

AI:
"""
        response = agent(
            prompt,
            max_new_tokens=60,
            temperature=0.7,
            top_k=50,
            top_p=0.95,
            repetition_penalty=1.2,
            do_sample=True,
            truncation=True,
            pad_token_id=50256
        )

        generated_text = response[0]["generated_text"]

        ai_reply = generated_text[len(prompt):]
        ai_reply = clean(ai_reply)

        if len(ai_reply) < 2:
            ai_reply = "I'm not sure how to answer that yet."

        memory.append(f"AI:  {ai_reply}")

        print("\n--------------------------------") # Fixed typo: /n to \n
        print("AI RESPONSE") # Corrected spelling: RSPONSE to RESPONSE
        print("----------------------------------")
        print(ai_reply)
        print("----------------------------------")

ai_agent()
```

```
from transformers import pipeline

agent = pipeline("text-generation", model="distilgpt2") #TinyLlama/TinyLlama-1.1B-chat-1.0
memory = []

def ai_agent():
  print("\n=== Simple AI Agent===\n ")
  print("Type '== Exit ==' to stop")

  while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
      print('Agent:  Goodbye!')
      break


    memory.append(user_input)
    history = "\n".join(memory[-5:])
    prompt = f"""
Conversation:
{history}

User Questions:
{user_input}
AI Response:
"""

    response = agent(prompt, max_new_tokens=150,top_k=50,top_p=0.95, repetition_penalty=1.2, do_sample=True, temperature=0.7)
    output = response[0]["generated_text"]
    air_reply = output[len(prompt):].strip()
    stop_tokens = ["User:", "conversation", "\n\n"]

    for token in stop_tokens:
      if token in air_reply:
        air_reply = air_reply.split(token)[0]

    ai_reply = air_reply.strip()

    memory.append(f"Agent: {ai_reply}")

    print("\nAgent: ")
    print(air_reply)
    print("\n")

ai_agent()
```
